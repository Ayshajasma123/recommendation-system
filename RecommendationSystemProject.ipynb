```python
#LOADING DATASET

import pandas as pd
import numpy as np
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
df=pd.read_csv("/content/archive (11).zip")
df
```

```python
#INSERTING INDEX COLUMN
df.insert(0, 'Index', range(1, len(df) + 1))
```

```python
df.head()
```

```python
#NO OF ROWS AND COLUMNS IN DATAFRAME

print("no of columns:",df.shape[1])
print("no of rows:",df.shape[0])
```

```python
#CHECKING MISSING VALUE

df.isna().sum()
```

```python
#CHECKING COLUMNS

df.columns
```

```python
#SELECTING RELEVENT FEATURES FOR RECOMMENDATION

selected_features=["name","price","corpus"]
print(selected_features)
```

```python
#REPLACING THE NULL VALUES WITH NULL STRINGS

for feature in selected_features:
  df[feature]=df[feature].fillna("")
```

```python
#COMBINING SELECTED FEATURES

combined_features=df["name"]+' '+df["price"]+' '+df["corpus"]
print(combined_features)
```

```python
#CONVERTING TEXT DATA TO VECTORS
vectorizer=TfidfVectorizer()
feature_vectors= vectorizer.fit_transform(combined_features)
print(feature_vectors)
```

```python
#GETTING SIMILARITY SCORES USING COSINE SIMILARITY
similarity=cosine_similarity(feature_vectors)
print(similarity)
```

```python
print(similarity.shape)
```

```python
#GETTING THE PHONE NAME FROM THE USER
mobile_name=input("enter your favourite phone name:")
```

```python
#CREATING A LIST WITH ALL THE MOBILE NAME GIVEN IN THE DATASET
list_of_all_name = df["name"].tolist()
print(list_of_all_name)
```

```python
#FINDING THE CLOSE MATCH MOBILE NAME GIVEN BY THE USER

find_close_match=difflib.get_close_matches(mobile_name,list_of_all_name)
print(find_close_match)
```

```python
close_match=find_close_match[0]
print(close_match)
```

```python
#FINDING THE INDEX OF THE MOBILE WITH NAME
index_of_the_mobile=df[df.name==close_match]["index"].values[0]
print(index_of_the_mobile)
```

```python
#GETTING A LIST OF SIMILIAR MOBILE
similarity_score=list(enumerate(similarity[index_of_the_mobile]))
print(similarity_score)
```

```python
len(similarity_score)
```

```python
#SORTING THE MOBILES BASED ON SIMILARITY SCORE
sorted_similar_mobiles=sorted(similarity_score, key=lambda x:x[1], reverse=True)
print(sorted_similar_mobiles)
```

```python
#PRINT SIMILAR MOBILES BASED ON THE INDEX
print("mobiles suggested for you: /n")
i=1
for mobile in sorted_similar_mobiles:
  index=mobiles[0]
  name_from_index=df[df.index==index]["title"].values[0]
  if (i<30):
    print(i,".",name_from_index)
    i+=1
```
